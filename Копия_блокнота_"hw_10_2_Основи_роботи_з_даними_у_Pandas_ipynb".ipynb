Відповідь напишіть текстом в окремій клітинці типу Markdown.

Це завдання - на продуктове мислення. Якщо не знаєте, які гіпотези сформулювати - напишіть ті, які здаються вам зовсім примітивними. Важливо навчитись висловлювати свої гіпотези щодо даних, і з часом та досвідом вони будуть ставати щоразу глибше і точніше. Весь аналіз даних будується на тому, що ми перевіряємо певні гіпотези та робимо висновки і рекоемндації на основі своїх досліджень.

```python
mean_yangon = df[df['City'] == 'Yangon']['gross income'].mean()
mean_mandalay = df[df['City'] == 'Mandalay']['gross income'].mean()

print(mean_yangon, mean_mandalay)
```

Гіпотеза 2 - більша середня кількість товарів у чеку = більше gross income
Гіпотеза 3 - більша середня ціна одиниці товару = більше gross income

```python
df[(df['City'] =='Yangon') | (df['City'] =='Mandalay')].groupby('City')['Quantity'].mean()
```

```python
df[(df['City'] =='Yangon') | (df['City'] =='Mandalay')].groupby('City')['Unit price'].mean()
```

Перевірка показала:
Гіпотеза 1 -  більша кількість чеків не дорівнює більше gross incomeщо
Гіпотеза 2, Гіпотеза 3 - середня кількість товарів у чеку (Quantity) та середня ціна одиниці товару (Unit price) практично однакові в обох містах (різниця менше 1%). Це означає, що ці фактори не є основною причиною різниці в середньому gross income
