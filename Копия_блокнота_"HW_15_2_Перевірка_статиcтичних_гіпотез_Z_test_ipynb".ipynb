1. Запишіть параметри задачі у змінні Python:
    - `sample_mean` - $\bar{x}$, середнє значення в вибірці
    - `population_mean` - $\mu_0$, середнє значення в популяції (тобто те, з яким ми порівнюємо середнє вибіркове значення)
    - `population_std` - $\sigma$, вибіркове стандартне відхилення, яке ми вважаємо, що є рівним ст. відх. популяції, адже маємо достатньо велику вибірку.
    - `sample_size` - $n$, розмір вибірки
    - `alpha` - $\alpha$ рівень значущості
    
    **Увага!** Для обчислення стандартного відхилення маємо скористатись функцією `np.std(your_dataframe, ddof=1)`. Чому так - розбираємо в лекції про t-test.


2. Визначте для цієї задачі:

    - якою є альтернативна гіпотеза $H_a$

3. Обчисліть стандартну помилку SE.
4. Розрахуйте z-статистику.

6. Обчисліть p-value з допомогою бібліотеки stats.
7. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$. Для прийняття рішення зробіть і порівняння z-статистики з критичним значенням, і проаналізуйте p-value.

```python
import numpy as np

sample_mean = np.mean(data)
population_std = np.std(data, ddof=1)
sample_size = len(data)

population_mean = 20
alpha = 0.05

print(f"Середнє у вибірці (sample_mean): {sample_mean:.4f}")
print(f"Стандартне відхилення (population_std): {population_std:.4f}")
print(f"Розмір вибірки (sample_size): {sample_size}")
```

* **Нульова гіпотеза ($H_0$):** $\mu = 20$ (середній чек не змінився і залишився на рівні \$20).
* **Альтернативна гіпотеза ($H_a$):** $\mu > 20$ (середній чек статистично значущо підвищився).
* **Тип тесту:** Оскільки в альтернативній гіпотезі стоїть знак $>$, ми маємо справу з **правостороннім тестом**.


```python
SE = population_std / np.sqrt(sample_size)

print(f"Стандартна помилка (SE): {SE:.4f}")
```

```python
z_stat = (sample_mean - population_mean) / SE

print(f"Фактична Z-статистика (z_stat): {z_stat:.4f}")
```

```python
from scipy import stats

# Критичне значення Z для правостороннього тесту (квантиль 1 - alpha)
z_crit = stats.norm.ppf(1 - alpha)

# p-value для правостороннього тесту (площа праворуч від нашого z_stat)
p_value = 1 - stats.norm.cdf(z_stat)

print(f"Критичне Z-значення (z_crit): {z_crit:.4f}")
print(f"Отримане p-value: {p_value:.4f}")
```

```python
print("ФІНАЛЬНИЙ ВИСНОВОК")
# Спосіб 1: Порівняння Z-статистики
if z_stat >= z_crit:
    print(f"Оскільки z_stat ({z_stat:.4f}) >= z_crit ({z_crit:.4f}) -> H0 ВІДХИЛЯЄТЬСЯ.")
else:
    print(f"Оскільки z_stat ({z_stat:.4f}) < z_crit ({z_crit:.4f}) -> H0 НЕ відхиляється.")

# Спосіб 2: Порівняння p-value
if p_value <= alpha:
    print(f"Оскільки p-value ({p_value:.4f}) <= alpha ({alpha}) -> Різниця статистично значуща.")
else:
    print(f"Оскільки p-value ({p_value:.4f}) > alpha ({alpha}) -> Немає підстав відхиляти H0.")
```

Виконайте обчислення z-test з використанням бібліотеки statsmodels.

Отримайте z-статистику та р-значення.

Виведіть p-значення та зробіть висновок, чи ми маємо достатньо доказів, аби стверджувати, що середній чек зріс.

Чи зійшлись значення z-статистику та р-значення в цьому завданні з попередніми обчисленнями?

```python
from statsmodels.stats.weightstats import ztest

# value=20 — це µ0
# alternative='larger' — вказує на правосторонній тест
z_stat_sm, p_value_sm = ztest(data, value=20, alternative='larger')

print("РЕЗУЛЬТАТИ STATSMODELS")
print(f"Z-статистика: {z_stat_sm:.4f}")
print(f"p-value: {p_value_sm:.4f}")
```

1. **Чи маємо ми достатньо доказів, аби стверджувати, що середній чек зріс?**
Ні, не маємо. Оскільки отримане $p\text{-value}$ значно більше за рівень значущості $\alpha = 0.05$ (воно становить близько $0.8402$), ми не маємо підстав відхиляти нульову гіпотезу. Зростання чека статистично не підтверджено.

2. **Чи зійшлися значення z-статистики та р-значення в цьому завданні з попередніми обчисленнями?**
Так, значення повністю збіглися. Автоматичний розрахунок через бібліотеку `statsmodels` видає точно такі самі цифри ($z \approx -0.9953$, $p \approx 0.8402$), як і ваші ручні покрокові обчислення з першого завдання. Це доводить, що обидва методи були реалізовані правильно.
