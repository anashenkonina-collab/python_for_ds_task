# Тести для малих вибірок та пропорцій



**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

Для цього виконайте наступні 6 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту.
2. Обчисліть вибіркові статистики: $\bar x$, $s$, $n$.
3. Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості $\alpha = 0.10$.
4. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$ використовуючи p-value.
5. Напишіть висновок: чи справді редизайн сайту подовжив час до покупки?
6. Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$.


```python
import numpy as np
from scipy import stats

data = np.array([128, 135, 121, 142, 126, 151, 123])
mu_0 = 123.7
```

```python
# 2. Обчислення вибіркових статистик
x_bar = np.mean(data)
s = np.std(data, ddof=1)
n = len(data)
df = n - 1  # ступені свободи

# 3. Обчислення стандартної помилки та t-статистики
SE = s / np.sqrt(n)
t_stat = (x_bar - mu_0) / SE

# Критичні значення для правостороннього тесту
t_crit_10 = stats.t.ppf(1 - 0.10, df)
t_crit_05 = stats.t.ppf(1 - 0.05, df)

# p-value для правостороннього тесту
p_value = 1 - stats.t.cdf(t_stat, df)

print("ВИБІРКОВІ СТАТИСТИКИ")
print(f"Середнє значення (x_bar): {x_bar:.2f}")
print(f"Стандартне відхилення (s): {s:.2f}")
print(f"Розмір вибірки (n): {n}")
print(f"Стандартна помилка (SE): {SE:.2f}")

print("РЕЗУЛЬТАТИ ТЕСТУ")
print(f"Фактична t-статистика: {t_stat:.2f}")
print(f"Критичне t (alpha=0.10): {t_crit_10:.2f}")
print(f"Критичне t (alpha=0.05): {t_crit_05:.2f}")
print(f"p-value: {p_value:.4f}")
```

1. **Гіпотези та тип тесту:**
   * Нульова гіпотеза ($H_0$): $\mu = 123.7$ хв (середній час до покупки після редизайну не змінився).
   * Альтернативна гіпотеза ($H_a$): $\mu > 123.7$ хв (середній час до покупки збільшився).
   * **Тип тесту:** Правосторонній одновибірковий t-тест (оскільки $n < 30$).

2. **Вибіркові статистики:**
   * Вибіркове середнє ($\bar{x}$) $\approx 132.29$
   * Вибіркове стандартне відхилення ($s$) $\approx 10.98$
   * Розмір вибірки ($n$) = $7$

3. **Результати тесту ($\alpha = 0.10$):**
   * Фактичне значення статистики: $t_{факт} = 2.07$
   * Критичне значення: $t_{крит} = 1.44$
   * Отримане $p\text{-value} = 0.0422$

4. **Прийняття рішення ($\alpha = 0.10$):**
   * Оскільки $p\text{-value} \ (0.0422) \le \alpha \ (0.10)$ або $t_{факт} \ (2.07) \ge t_{крит} \ (1.44)$, фактичне значення потрапляє в область відхилення. **Гіпотеза $H_0$ відхиляється.**

5. **Висновок для бізнесу ($\alpha = 0.10$):**
   * Так, редизайн сайту статистично значущо продовжив середній час до покупки на рівні значущості 10%.

6. **Вплив зміни рівня значущості на ($\alpha = 0.05$):**
   * При $\alpha = 0.05$ критичне значення становить $t_{крит} = 1.94$.
   * Наше $t_{факт} = 2.07$ все одно залишається більшим за нове критичне значення ($2.07 \ge 1.94$), а $p\text{-value} \ (0.0422) \le 0.05$.
   * **Рішення не зміниться:** Нульова гіпотеза $H_0$ так само відхиляється. Результат є стійким навіть при суворішому критерії перевірки.


**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest, proportion_confint

#Вихідні дані з умови
count = 311        # кількість тих, хто впізнав логотип
nobs = 1200       # загальна кількість опитаних
p_0 = 0.23        # історична частка (23%)
alpha = 0.03      # рівень значущості (3%)

# Z-тест для часток
z_stat, p_value = proportions_ztest(count, nobs, value=p_0, alternative='larger')

# довірчий інтервал (1 - alpha = 0.97)
ci_low, ci_upp = proportion_confint(count, nobs, alpha=alpha, method='normal')

print("РЕЗУЛЬТАТИ Z-ТЕСТУ")
print(f"Вибіркова частка (p_hat): {count/nobs:.4f}")
print(f"Z-статистика: {z_stat:.4f}")
print(f"p-value: {p_value:.4f}")

print("ДОВІРЧИЙ ІНТЕРВАЛ (97%)")
print(f"Нижня межа: {ci_low:.4f} ({ci_low*100:.2f}%)")
print(f"Верхня межа: {ci_upp:.4f} ({ci_upp*100:.2f}%)")

```


1. **Формулювання гіпотез:**
   * Нульова гіпотеза ($H_0$): $p = 0.23$ (впізнаваність логотипу не змінилася і залишилася на рівні 23%).
   * Альтернативна гіпотеза ($H_a$): $p > 0.23$ (впізнаваність логотипу після рекламної кампанії зросла).
   * **Рівень значущості:** $\alpha = 0.03$.

2. **Результати обчислень:**
   * Вибіркова частка: $\hat{p} = \frac{311}{1200} \approx 0.2592$ (або $25.92\%$).
   * Фактичне значення критерію: $Z = 2.4009$
   * Отримане значення $p\text{-value} = 0.0082$

3. **Прийняття рішення:**
   * Оскільки $p\text{-value} \ (0.0082) \le \alpha \ (0.03)$, ми маємо вагомі підстави **відхилити нульову гіпотезу ($H_0$)**.

4. **Висновок про впізнаваність:**
   * Рекламна кампанія була успішною. На рівні значущості 3% дані дають достатньо підстав стверджувати, що впізнаваність логотипу компанії статистично значущо **зросла** і тепер становить більше ніж 23%.

5. **Інтерпретація довірчого інтервалу:**
   * Побудований 97%-й довірчий інтервал для нової частки становить від **23.17% до 28.66%**.
   *  Оскільки базове історичне значення ($23\%$) лежить повністю *поза межами* цього інтервалу (воно менше навіть за найнижчу очікувану межу в 23.17%), це додатково підтверджує наш висновок. Ми можемо бути на 97% впевнені, що справжній новий рівень впізнаваності перевищує старий показник.
